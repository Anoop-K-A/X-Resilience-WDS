```python
# Create an observation-segment ID.
# A new segment starts whenever the gap between observations is > 1 hour.

time_diff = df["DATETIME"].diff()

df["SEGMENT_ID"] = (
    time_diff.gt(pd.Timedelta(hours=1))
    .cumsum()
)

print("Number of observation segments:", df["SEGMENT_ID"].nunique())

print("\nSegment summary:")
segment_summary = (
    df.groupby("SEGMENT_ID")
      .agg(
          start=("DATETIME", "min"),
          end=("DATETIME", "max"),
          rows=("DATETIME", "size"),
          attacks=("ATT_FLAG", lambda x: (x == 1).sum())
      )
      .reset_index()
)

print(segment_summary.to_string(index=False))
```

```python
# Create attack-event IDs.
# An attack event starts when:
# 1. ATT_FLAG changes from non-attack to attack
# 2. OR there is a time gap greater than 1 hour

is_attack = df["ATT_FLAG"] == 1
new_attack = (
    is_attack
    & (
        ~is_attack.shift(fill_value=False)
        | df["SEGMENT_ID"].ne(df["SEGMENT_ID"].shift())
    )
)

df["ATTACK_EVENT_ID"] = new_attack.cumsum()

# Give non-attack rows no event ID
df.loc[~is_attack, "ATTACK_EVENT_ID"] = 0

# Summarize attack events
attack_events = (
    df[df["ATT_FLAG"] == 1]
    .groupby("ATTACK_EVENT_ID")
    .agg(
        segment=("SEGMENT_ID", "first"),
        start=("DATETIME", "min"),
        end=("DATETIME", "max"),
        rows=("DATETIME", "size")
    )
    .reset_index()
)

print(attack_events.to_string(index=False))
```

```python
import numpy as np

RANDOM_SEED = 42
rng = np.random.default_rng(RANDOM_SEED)

# Attack event information
events = attack_events[
    ["ATTACK_EVENT_ID", "rows"]
].copy()

event_ids = events["ATTACK_EVENT_ID"].to_numpy()

best_split = None
best_score = float("inf")

# Try many random event assignments and keep the most balanced one
for _ in range(10000):

    shuffled = rng.permutation(event_ids)

    train_ids = shuffled[:6]
    val_ids = shuffled[6:8]
    test_ids = shuffled[8:11]

    train_rows = events.loc[
        events["ATTACK_EVENT_ID"].isin(train_ids), "rows"
    ].sum()

    val_rows = events.loc[
        events["ATTACK_EVENT_ID"].isin(val_ids), "rows"
    ].sum()

    test_rows = events.loc[
        events["ATTACK_EVENT_ID"].isin(test_ids), "rows"
    ].sum()

    # We want validation and test to have reasonably similar
    # numbers of attack observations.
    score = abs(val_rows - test_rows)

    if score < best_score:
        best_score = score
        best_split = (train_ids, val_ids, test_ids)

train_events, val_events, test_events = best_split

print("Training events:   ", train_events)
print("Validation events: ", val_events)
print("Test events:       ", test_events)

print("\nAttack rows:")
print(
    "Train:      ",
    events.loc[events["ATTACK_EVENT_ID"].isin(train_events), "rows"].sum()
)
print(
    "Validation: ",
    events.loc[events["ATTACK_EVENT_ID"].isin(val_events), "rows"].sum()
)
print(
    "Test:       ",
    events.loc[events["ATTACK_EVENT_ID"].isin(test_events), "rows"].sum()
)
```

```python
# ============================================================
# Temporal feature engineering
# ============================================================

import numpy as np
import pandas as pd

# Continuous physical measurements
continuous_cols = pressure_cols + flow_cols + tank_cols

# Status measurements
status_cols = status_cols


def create_temporal_features(data):
    """
    Create real-time-safe temporal features.

    Features use only the current observation and past observations.
    Calculations are reset at each continuous observation segment.
    """

    result = data.copy()

    # Make sure chronological order is preserved
    result = result.sort_values(
        ["SEGMENT_ID", "DATETIME"]
    ).copy()

    # --------------------------------------------------------
    # Continuous sensor features
    # --------------------------------------------------------

    for col in continuous_cols:

        # Change from previous observation
        result[f"{col}_delta1"] = (
            result.groupby("SEGMENT_ID")[col]
            .diff(1)
        )

        # Recent 3-hour behaviour
        result[f"{col}_rollmean3"] = (
            result.groupby("SEGMENT_ID")[col]
            .transform(
                lambda x: x.rolling(
                    window=3,
                    min_periods=1
                ).mean()
            )
        )

        result[f"{col}_rollstd3"] = (
            result.groupby("SEGMENT_ID")[col]
            .transform(
                lambda x: x.rolling(
                    window=3,
                    min_periods=1
                ).std()
            )
        )

    # --------------------------------------------------------
    # Status change features
    # --------------------------------------------------------

    for col in status_cols:

        result[f"{col}_changed"] = (
            result.groupby("SEGMENT_ID")[col]
            .diff(1)
            .abs()
            .fillna(0)
        )

    return result


feature_df = create_temporal_features(df)

print("Original columns:", len(df.columns))
print("Feature dataframe columns:", len(feature_df.columns))

print("\nNew temporal features:")
new_features = [
    c for c in feature_df.columns
    if c not in df.columns
]

print(new_features[:30])
print("...")
print("Total new features:", len(new_features))
```

```python
# ============================================================
# Temporal feature engineering - optimized version
# ============================================================

import numpy as np
import pandas as pd


continuous_cols = pressure_cols + flow_cols + tank_cols


def create_temporal_features(data):
    """
    Create real-time-safe temporal features.

    All temporal calculations use only the current observation
    and previous observations within the same continuous segment.
    """

    base = data.sort_values(
        ["SEGMENT_ID", "DATETIME"]
    ).copy()

    feature_parts = [base]

    # --------------------------------------------------------
    # Continuous sensor features
    # --------------------------------------------------------

    temporal_features = {}

    for col in continuous_cols:

        grouped = base.groupby("SEGMENT_ID")[col]

        temporal_features[f"{col}_delta1"] = (
            grouped.diff(1)
        )

        temporal_features[f"{col}_rollmean3"] = (
            grouped.transform(
                lambda x: x.rolling(
                    window=3,
                    min_periods=1
                ).mean()
            )
        )

        temporal_features[f"{col}_rollstd3"] = (
            grouped.transform(
                lambda x: x.rolling(
                    window=3,
                    min_periods=1
                ).std()
            )
        )

    # --------------------------------------------------------
    # Status change features
    # --------------------------------------------------------

    for col in status_cols:

        grouped = base.groupby("SEGMENT_ID")[col]

        temporal_features[f"{col}_changed"] = (
            grouped.diff(1)
            .abs()
            .fillna(0)
        )

    # Create all new columns at once
    temporal_df = pd.DataFrame(
        temporal_features,
        index=base.index
    )

    result = pd.concat(
        [base, temporal_df],
        axis=1
    )

    return result


feature_df = create_temporal_features(df)

print("Original columns:", len(df.columns))
print("Feature dataframe columns:", len(feature_df.columns))

new_features = [
    c for c in feature_df.columns
    if c not in df.columns
]

print("\nTotal new features:", len(new_features))
print("\nFirst 30 new features:")
print(new_features[:30])
```
